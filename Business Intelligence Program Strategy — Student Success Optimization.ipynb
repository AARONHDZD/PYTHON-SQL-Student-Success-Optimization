```python
# Import the raw dataset and inspect its structure,
# columns, and overall information.

import pandas as pd

data = pd.read_csv("bi.csv", encoding='latin-1')
data.head()
data.tail()
data.shape
# The dataset contains 77 rows and 10 columns.
```

```python
# Identify missing values across the dataset.
data.isnull().sum()

# The Python column contains 2 missing values.
# These values will be replaced using the column mean.
data['Python'] = data['Python'].fillna(round(data['Python'].mean(),2))
data.head()
data.isnull().sum()

# Check for duplicate records.
data.duplicated().sum()
```

```python
# Identify inconsistent categorical values across string-based columns.
# Standardize categorical values for Gender, Country, 
# Residence, and Previous Education.

data.gender.unique() #'Female', 'M', 'Male', 'F', 'female', 'male'
genderstd = {'M':'Male',
             'F': 'Female',
             'female':'Female',
             'male': 'Male'}
data['gender'] = data['gender'].replace(genderstd)
data['gender'].unique()

data['country'].unique()
countrystd = {'Rsa':'South Africa',
              'norway':'Norway',
              'Norge':'Norway',
              'UK':'United Kingdom',
              'Somali':'Somalia'}
data['country'] = data['country'].replace(countrystd)
data['country'].unique()

data['residence'].unique() # 'Private',    'Sognsvann', 'BI Residence', 'BI-Residence', 'BIResidence', 'BI_Residence'
residstd = {'BI-Residence':'BI Residence',
            'BIResidence':'BI Residence',
            'BI_Residence':'BI Residence'}
data['residence'] = data['residence'].replace(residstd)
data['residence'].unique()

data['prevEducation'].unique() # 'Masters',  'Diploma',  'HighSchool',  'High School', 'Bachelors', 'Barrrchelors','diploma', 'DIPLOMA', 'Diplomaaa', 'Doctorate'
educstd = {'diploma':'Diploma',
           'Diplomaaa':'Diploma',
           'DIPLOMA':'Diploma',
           'HighSchool':'High School',
           'Barrrchelors':'Bachelors'}
data['prevEducation'] = data['prevEducation'].replace(educstd)
data['prevEducation'].unique()
```

```python
# Convert score-related columns from integer
# to float data types for consistency.

data.dtypes
data['entryEXAM'] = data['entryEXAM'].astype('float64')
data['Python'] = data['Python'].astype('float64')
data['DB'] = data['DB'].astype('float64')
data.dtypes
```

```python
# Rename columns using a cleaner naming convention
# and create a FullName column for better readability.

data.columns #fNAME', 'lNAME', 'Age', 'gender', 'country', 'residence', 'entryEXAM', 'prevEducation', 'studyHOURS', 'Python', 'DB'
data = data.rename(columns={'fNAME':'FirstName',
    'lNAME':'LastName',
    'gender':'Gender',
    'country':'Country',
    'residence':'Residence',
    'entryEXAM':'EntryExam',
    'prevEducation':'PrevEducation',
    'studyHOURS':'StudyHours'})
data['FullName'] = data['FirstName'] + ' ' + data['LastName']
data.insert(2,'FullName', data.pop('FullName'))

data.head()
```

```python
# Before exporting the cleaned dataset for SQL analysis,
# identify and handle potential outliers.

# Validate potential outliers in Age and score-related columns.
# Expected age values should range between 20 and 60 years,
# while score values should remain within valid academic ranges.

data['Age'].sort_values() #Index 55 and 32 must be replaced with average age
data.loc[[55,32], 'Age']=data['Age'].mean().astype('int')
data.loc[55, 'Age']

data['EntryExam'].sort_values() # No anomalies detected.
data['Python'].sort_values() # No anomalies detected.
data['DB'].sort_values() # No anomalies detected.
data
```
